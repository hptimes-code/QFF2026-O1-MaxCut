```python
from pathlib import Path
import subprocess, sys
ROOT = Path.cwd()
assert (ROOT / "main.py").exists(), "Start Jupyter in the repository root"
def run(*args):
    subprocess.run([sys.executable, *map(str,args)], cwd=ROOT, check=True)
```
